# Data inspecting, editing and cleaning
'''
Map company_id to a continuous index (0,1,2...) for your stock embeddings since the model needs sequential IDs.

Check if timestamps are regular (e.g., every 1min) – look for gaps to decide if you need time-based positional encoding later.

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('test.db')

df = pd.read_sql_query("SELECT * FROM stock_prices", conn)

conn.close()

print(df.info())
```

```python
#mapping exists
print(df['company_id'].unique())
```

```python
#data types change
df['date'] = pd.to_datetime(df['date'])
df['timestamp'] = pd.Categorical(df['timestamp'])

df.info()
```

```python
# Filter for first company and 1m timestamp
company1_1m = df[(df['company_id'] == 1) & (df['timestamp'] == '1m')].sort_values('date').copy()

# Calculate rolling 10-period percentage change (row 11 vs row 1, row 12 vs row 2, etc.)
company1_1m['close_price_rolling_10period_pct_change'] = (company1_1m['close_price'] / company1_1m['close_price'].shift(10) - 1) * 100

# Display results
print(f"Data for Company 1, 1m interval: {len(company1_1m)} rows")
print("\nRolling 10-period percentage changes (current vs 10 rows ago):")
result = company1_1m[['date', 'close_price', 'close_price_rolling_10period_pct_change']].copy()
print(result.head(20))

print(company1_1m[['date', 'close_price', 'close_price_rolling_10period_pct_change']].sort_values('close_price_rolling_10period_pct_change', ascending=False).dropna().tail(3))
```

```python
# Calculate percentage changes
company1_1m['close_price_1period_pct_change'] = company1_1m['close_price'].pct_change() * 100

# Create 5 categories using quantile-based splitting
company1_1m['change_category'] = pd.qcut(company1_1m['close_price_1period_pct_change'], 
                                         q=5, 
                                         labels=['very low', 'low', 'medium', 'high', 'very high'],
                                         duplicates='drop')

# Sort and display
company1_1m_sorted = company1_1m[['date', 'close_price', 'close_price_1period_pct_change', 'change_category']].sort_values('close_price_1period_pct_change', ascending=False)

print("Top 10 largest positive changes:")
print(company1_1m_sorted.head(10))

print("\nBottom 10 largest negative changes:")
print(company1_1m_sorted.tail(10))

# Show distribution of categories (should be roughly equal now)
print("\nCategory distribution:")
print(company1_1m['change_category'].value_counts().sort_index())

company1_1m.groupby('change_category', observed=True)['close_price_1period_pct_change'].agg(['mean', 'min', 'max', 'count'])
```

```python
company1_1m['min5PctChange'] = (company1_1m['close_price'] / company1_1m['close_price'].shift(5) - 1) * 100
company1_1m['positive_peak_5min'] = (company1_1m['min5PctChange'] > 0.5) & (company1_1m['min5PctChange'].notna())
company1_1m['negative_peak_5min'] = (company1_1m['min5PctChange'] < -0.5) & (company1_1m['min5PctChange'].notna())

print(company1_1m['min5PctChange'].tail())
print(company1_1m['positive_peak_5min'].value_counts())
print(company1_1m['negative_peak_5min'].value_counts())
```

```python
# Separate positive and negative peaks
positive_peaks = company1_1m[company1_1m['positive_peak_5min']][['date', 'min5PctChange']].sort_values('date').copy()
negative_peaks = company1_1m[company1_1m['negative_peak_5min']][['date', 'min5PctChange']].sort_values('date').copy()

# Suppression function
def suppress_peaks(peaks_df, minutes=5):
    keep_indices = []
    i = 0
    while i < len(peaks_df):
        current_idx = peaks_df.index[i]
        current_date = peaks_df.iloc[i]['date']
        current_value = peaks_df.iloc[i]['min5PctChange']
        keep_indices.append(current_idx)
        
        # Look ahead for peaks within time window
        j = i + 1
        while j < len(peaks_df):
            next_date = peaks_df.iloc[j]['date']
            time_diff = (next_date - current_date).total_seconds() / 60
            if time_diff <= minutes:
                # Skip this peak (don't add to keep_indices)
                j += 1
            else:
                break
        i = j
    
    return keep_indices

# Apply suppression separately
positive_keep = suppress_peaks(positive_peaks)
negative_keep = suppress_peaks(negative_peaks)

# Create final flags
company1_1m['final_positive_peak'] = False
company1_1m['final_negative_peak'] = False
company1_1m.loc[positive_keep, 'final_positive_peak'] = True
company1_1m.loc[negative_keep, 'final_negative_peak'] = True

print(f"Original positive peaks: {len(positive_peaks)} -> Final: {len(positive_keep)}")
print(f"Original negative peaks: {len(negative_peaks)} -> Final: {len(negative_keep)}")
```
